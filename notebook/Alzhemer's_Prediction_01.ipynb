```python
#importing datasetimport pandas as pd

import pandas as pd

df = pd.read_csv("data/alzheimers_disease_data.csv")

print("Shape:", df.shape)
df.head()
```

```python
print("Missing values per column:\n", df.isnull().sum()[df.isnull().sum() > 0])
print("\nDiagnosis value counts:\n", df["Diagnosis"].value_counts())
print("\nColumns:\n", list(df.columns))
```

```python
#dropping unecessary columns


df_clean = df.drop(columns=["PatientID", "DoctorInCharge"])


X = df_clean.drop(columns=["Diagnosis"])
y = df_clean["Diagnosis"]

print("X shape:", X.shape)
print("y shape:", y.shape)
X.dtypes
```

```python
Feature Engineering
```

```python
X_fe = X.copy()

# Blood pressure: pulse pressure often carries more signal than either value alone
X_fe["PulsePressure"] = X_fe["SystolicBP"] - X_fe["DiastolicBP"]

# Cholesterol ratio: LDL/HDL is a standard cardiovascular risk indicator
X_fe["CholesterolRatio"] = X_fe["CholesterolLDL"] / X_fe["CholesterolHDL"]

# Symptom burden: sum of binary cognitive/behavioral symptom flags
symptom_cols = [
    "MemoryComplaints", "BehavioralProblems", "Confusion",
    "Disorientation", "PersonalityChanges",
    "DifficultyCompletingTasks", "Forgetfulness"
]
X_fe["SymptomCount"] = X_fe[symptom_cols].sum(axis=1)

# Comorbidity burden: sum of binary medical history flags
comorbidity_cols = [
    "FamilyHistoryAlzheimers", "CardiovascularDisease",
    "Diabetes", "Depression", "HeadInjury", "Hypertension"
]
X_fe["ComorbidityCount"] = X_fe[comorbidity_cols].sum(axis=1)

# Composite cognitive/functional decline score (MMSE and FunctionalAssessment/ADL
# are all scaled so that LOWER = more impaired, so we can combine them directly)
X_fe["CognitiveFunctionalScore"] = X_fe["MMSE"] + X_fe["FunctionalAssessment"] + X_fe["ADL"]

print("New shape:", X_fe.shape)
X_fe[["PulsePressure", "CholesterolRatio", "SymptomCount", "ComorbidityCount", "CognitiveFunctionalScore"]].describe()
```

```python
X_fe
```

```python
#assing new column for bp categories

def bp_category(row):
    sys, dia = row["SystolicBP"], row["DiastolicBP"]
    
    if sys > 180 or dia > 120:
        return "Hypertensive Crisis"
    elif sys >= 140 or dia >= 90:
        return "Stage 2 Hypertension"
    elif sys >= 130 or dia >= 80:
        return "Stage 1 Hypertension"
    elif 120 <= sys <= 129 and dia < 80:
        return "Elevated"
    elif sys < 120 and dia < 80:
        return "Normal"
    else:
        
        return "Stage 1 Hypertension"

X_fe["BPCategory"] = X_fe.apply(bp_category, axis=1)
print(X_fe["BPCategory"].value_counts())
```

```python
#encoding bp

bp_order = {
    "Normal": 0,
    "Elevated": 1,
    "Stage 1 Hypertension": 2,
    "Stage 2 Hypertension": 3,
    "Hypertensive Crisis": 4
}
X_fe["BPCategoryEncoded"] = X_fe["BPCategory"].map(bp_order)

# Drop the text version now that we have the encoded one, and drop the raw
# PulsePressure calc's redundant inputs is NOT needed — keep SystolicBP/DiastolicBP too,
# since the raw values still carry information the category buckets lose
X_fe = X_fe.drop(columns=["BPCategory"])

print("New shape:", X_fe.shape)
X_fe[["SystolicBP", "DiastolicBP", "BPCategoryEncoded"]].head(10)
```

```python
# Map numeric codes to readable labels first (for clarity and to catch mapping errors)
gender_map = {0: "Male", 1: "Female"}
ethnicity_map = {0: "Caucasian", 1: "African American", 2: "Asian", 3: "Other"}

X_fe["GenderLabel"] = X_fe["Gender"].map(gender_map)
X_fe["EthnicityLabel"] = X_fe["Ethnicity"].map(ethnicity_map)

print(X_fe["GenderLabel"].value_counts())
print(X_fe["EthnicityLabel"].value_counts())
```

```python
#one hot encoding ggender

X_fe = pd.get_dummies(
    X_fe,
    columns=["GenderLabel", "EthnicityLabel"],
    prefix=["Gender", "Ethnicity"],
    drop_first=True  
)

# Drop the original numeric-coded columns now that we have the one-hot versions
X_fe = X_fe.drop(columns=["Gender", "Ethnicity"])

print("New shape:", X_fe.shape)
X_fe.columns.tolist()
```

```python
X_fe
```

```python
#ewncoding bmi

def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25.0:
        return "Healthy"
    elif bmi < 30.0:
        return "Overweight"
    elif bmi < 35.0:
        return "Obese Class 1"
    elif bmi < 40.0:
        return "Obese Class 2"
    else:
        return "Obese Class 3"

X_fe["BMICategory"] = X_fe["BMI"].apply(bmi_category)
print(X_fe["BMICategory"].value_counts())
```

```python
bmi_order = {
    "Underweight": 0,
    "Healthy": 1,
    "Overweight": 2,
    "Obese Class 1": 3,
    "Obese Class 2": 4,
    "Obese Class 3": 5
}
X_fe["BMICategoryEncoded"] = X_fe["BMICategory"].map(bmi_order)
X_fe = X_fe.drop(columns=["BMICategory"])

print("New shape:", X_fe.shape)
X_fe[["BMI", "BMICategoryEncoded"]].head(10)
```

```python
#categorizing cholestrol

def total_chol_category(val):
    if val < 200:
        return 0  # Desirable
    elif val < 240:
        return 1  # Borderline High
    else:
        return 2  # High

def ldl_category(val):
    if val < 100:
        return 0  # Optimal
    elif val < 130:
        return 1  # Near/Above optimal
    elif val < 160:
        return 2  # Borderline high
    elif val < 190:
        return 3  # High
    else:
        return 4  # Very high

def hdl_category(val, is_female):
    # Lower thresholds for men, higher for women
    low_cut = 50 if is_female else 40
    if val < low_cut:
        return 0  # Low risk / undesirable
    elif val <= 80:
        return 1  # Normal / acceptable 
    else:
        return 2  # High (above 80)

def triglyceride_category(val):
    if val < 150:
        return 0  # Normal
    elif val < 200:
        return 1  # Borderline high
    elif val < 500:
        return 2  # High
    else:
        return 3  # Very high

X_fe["TotalCholCategory"] = X_fe["CholesterolTotal"].apply(total_chol_category)
X_fe["LDLCategory"] = X_fe["CholesterolLDL"].apply(ldl_category)
X_fe["HDLCategory"] = X_fe.apply(lambda row: hdl_category(row["CholesterolHDL"], row["Gender_Female"] == 1), axis=1)
X_fe["TriglycerideCategory"] = X_fe["CholesterolTriglycerides"].apply(triglyceride_category)

print(X_fe[["TotalCholCategory", "LDLCategory", "HDLCategory", "TriglycerideCategory"]].apply(pd.Series.value_counts))
```

```python
#Train/test split


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_fe, y, test_size=0.2, random_state=42, stratify=y
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
#Featue Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)  
X_test_scaled = scaler.transform(X_test)          

print("Scaled train shape:", X_train_scaled.shape)
print("Scaled test shape:", X_test_scaled.shape)
```

```python
#fit_transform on train vs just transform on test — that asymmetry is the whole point, it's what prevents the leakage.
```

```python
#Random forest

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced"  # compensating for class imbalance
)

```

```python
rf_model.fit(X_train_scaled, y_train)
```

```python
#random forest metrics
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report, confusion_matrix
)

preds = rf_model.predict(X_test_scaled)
probs = rf_model.predict_proba(X_test_scaled)[:, 1]

print("Accuracy:", accuracy_score(y_test, preds))
print("Precision:", precision_score(y_test, preds))
print("Recall:", recall_score(y_test, preds))
print("F1 Score:", f1_score(y_test, preds))
print("ROC-AUC:", roc_auc_score(y_test, probs))

print("\nClassification Report:\n", classification_report(y_test, preds))
print("Confusion Matrix:\n", confusion_matrix(y_test, preds))
```

```python
#21 real AD patients (out of 152) got predicted as "No AD."
#We need tro reduce that number
```
